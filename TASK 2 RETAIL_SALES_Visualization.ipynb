```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme()
```

## 1. Load data
Upload `retail_sales_dataset.csv` to the Colab session before running.

```python
df = pd.read_csv('/retail_sales_dataset.csv')
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df.head()
```

## 5. Visualizations

```python
monthly = df.groupby(df['Order_Date'].dt.to_period('M'))['Sales_INR'].sum()
monthly.index = monthly.index.astype(str)
monthly.plot(marker='o', figsize=(9,5), title='Monthly Retail Sales')
plt.xlabel('Month'); plt.ylabel('Sales (INR)'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

```python
df.groupby('Category')['Sales_INR'].sum().sort_values().plot(kind='barh', figsize=(8,5), title='Sales by Category')
plt.xlabel('Sales (INR)'); plt.tight_layout(); plt.show()
```

```python
df.groupby('Region')['Sales_INR'].sum().plot(kind='bar', figsize=(7,5), title='Sales by Region')
plt.ylabel('Sales (INR)'); plt.xticks(rotation=0); plt.tight_layout(); plt.show()
```

```python
df['Payment_Method'].value_counts().plot(kind='pie', autopct='%1.1f%%', figsize=(7,7), title='Orders by Payment Method')
plt.ylabel(''); plt.show()
```
